```python
# Fill missing values

# Children: assume no children
df['children'] = df['children'].fillna(0)

# City: unknown city
df['city'] = df['city'].fillna('Unknown')

# Agent and company: no agent/company associated
df['agent'] = df['agent'].fillna('No Agent')
df['company'] = df['company'].fillna('No Company')
```

```python
df = df.dropna()
```

```python
#dropping duplicates
df.drop_duplicates(inplace=True)
```

**Feature Engineering**

```python
# 1. TOTAL STAY NIGHTS
df['total_stay_nights'] =(
    df['stays_in_weekend_nights'] +
    df['stays_in_weekdays_nights'])
```

```python
# 2. TOTAL GUESTS
df['total_guests'] = (
    df['adults'] +
    df['children'] +
    df['babies'])
```

```python
# 3. TOTAL PEOPLE NIGHTS
df['guest_nights'] = (
    df['total_guests'] * df['total_stay_nights']
)
```

```python
# 4. BOOKING PARTY TYPE
df['guest_type'] = np.where(
    df['total_guests'] == 1,
    'Single',
    np.where(
        df['total_guests'] == 2,
        'Couple',
        'Group'
    )
)
```

```python
# 5. LONG STAY / SHORT STAY
df['stay_category'] = pd.cut(
    df['total_stay_nights'],
    bins=[-1, 2, 5, 10, float('inf')],
    labels=['Short Stay', 'Medium Stay', 'Long Stay', 'Very Long Stay']
)
```

```python
# 6. LEAD TIME CATEGORY
df['lead_time_category'] = pd.cut(
    df['lead_time'],
    bins=[-1, 7, 30, 90, 180, float('inf')],
    labels=[
        'Last Minute',
        'Short Notice',
        'Medium Advance',
        'Long Advance',
        'Very Long Advance'
    ]
)
```

```python
# 7. REPEATED GUEST STATUS
df['guest_status'] = np.where(
    df['is_repeated_guest'] == 1,
    'Returning Guest',
    'New Guest'
)
```

```python
#Hotel booking volume by hotel type
hotel_counts = df["hotel"].value_counts()

ax = hotel_counts.plot(kind="bar", figsize=(9, 5))
ax.set_title("Booking Volume by Hotel Type")
ax.set_xlabel("Hotel Type")
ax.set_ylabel("Number of Bookings")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

City Hotel has about 52,500 bookings and Resort about 33,500, so City is roughly 61% of volume.

```python
# Monthly booking trend by hotel type (line chart)
monthly = (
    df.groupby(["arrival_date_month", "hotel"])
      .size()
      .reset_index(name="bookings")
)
monthly["month_num"] = monthly["arrival_date_month"].map({m:i for i,m in enumerate(month_order, 1)})
monthly = monthly.sort_values("month_num")

plt.figure(figsize=(12, 6))
sns.lineplot(
    data=monthly,
    x="arrival_date_month",
    y="bookings",
    hue="hotel",
    marker="o"
)
plt.title("Monthly Booking Trend by Hotel Type")
plt.xlabel("Arrival Month")
plt.ylabel("Number of Bookings")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Bookings are low in January–March (City about 2,700–3,000, Resort about 1,900–2,100).


They climb from April and peak in Sep–Oct (City about 6,500 and Resort about 4,650 in October).


They drop in Nov–Dec.


Resort jumps more sharply from August (about 2,700) to September (about 4,300).
Add the month-label caveat from above here.

```python
# Cancellation rate by hotel type (donut/pie chart)
cancel_by_hotel = df.groupby("hotel")["is_canceled"].mean() * 100

plt.figure(figsize=(7, 7))
plt.pie(
    cancel_by_hotel.values,
    labels=[f"{k}: {v:.1f}%" for k, v in cancel_by_hotel.items()],
    startangle=90,
    wedgeprops={"width": 0.42, "edgecolor": "white"}
)
plt.title("Cancellation Rate by Hotel Type")
plt.tight_layout()
plt.show()
```

City cancels about 30.2% and Resort about 23.5%, a gap of roughly 7 points. Overall it's about 28%, so more than 1 in 4 bookings is lost.

```python
# Cancellation rate by stay duration (grouped bar)
stay_cancel = (
    df.groupby(["hotel", "stay_category"], observed=True)["is_canceled"]
      .mean()
      .mul(100)
      .reset_index()
)

plt.figure(figsize=(11, 6))
sns.barplot(
    data=stay_cancel,
    x="stay_category",
    y="is_canceled",
    hue="hotel"
)
plt.title("Cancellation Rate by Stay Duration")
plt.xlabel("Stay Category")
plt.ylabel("Cancellation Rate (%)")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

City Hotel's rate rises with stay length: about 26%, 31.5%, 39%, then 64% for very long stays.

Resort stays fairly flat at about 15–29%, dipping to about 24.5% for very long stays.

```python
#Lead time distribution by cancellation status (histogram)
plt.figure(figsize=(11, 6))
sns.histplot(
    data=df,
    x="lead_time",
    hue="is_canceled",
    bins=40,
    kde=True,
    element="step",
    stat="density",
    common_norm=False
)
plt.title("Lead Time Distribution: Cancelled vs Non-Cancelled")
plt.xlabel("Lead Time (Days)")
plt.ylabel("Density")
plt.xlim(0, df["lead_time"].quantile(0.99))
plt.tight_layout()
plt.show()
```

```python
# Lead-time category cancellation rate (horizontal bar)
lead_cancel = (
    df.groupby(["hotel", "lead_time_category"], observed=True)["is_canceled"]
      .mean()
      .mul(100)
      .reset_index()
)

plt.figure(figsize=(11, 7))
sns.barplot(
    data=lead_cancel,
    y="lead_time_category",
    x="is_canceled",
    hue="hotel"
)
plt.title("Cancellation Rate by Booking Lead-Time Category")
plt.xlabel("Cancellation Rate (%)")
plt.ylabel("Lead-Time Category")
plt.tight_layout()
plt.show()
```

Non-cancelled bookings spike at under about 20 days, while cancelled ones have a heavier tail beyond about 25 days.

The earlier someone books, the more likely they cancel.
City goes from about 10.5% (0–7 days) through 28%, 33.5% and 37% to about 46% (180+ days).

Resort goes from about 6% through 21%, 30% and 32% to 34%.

```python
# ADR distribution by hotel type (box plot)
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x="hotel", y="adr")
plt.title("Average Daily Rate Distribution by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("ADR")
plt.ylim(0, df["adr"].quantile(0.99))
plt.tight_layout()
plt.show()
```

Resort's median is about 80, with a wide spread (roughly 52–135), which fits seasonal pricing.

City's median is about 105, with a tighter spread and many high outliers.
The near-zero values are likely free or complimentary bookings.

The chart is clipped at the 99th percentile.

```python
#Correlation heatmap of key numeric business variables
numeric_cols = [
    "is_canceled", "lead_time", "total_stay_nights", "total_guests",
    "guest_nights", "previous_cancellations", "previous_bookings_not_canceled",
    "booking_changes", "days_in_waiting_list", "adr",
    "required_car_parking_spaces", "total_of_special_requests"
]
numeric_cols = [c for c in numeric_cols if c in df.columns]

corr = df[numeric_cols].corr()

plt.figure(figsize=(13, 9))
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)
plt.title("Correlation Heatmap — Key Booking & Cancellation Variables")
plt.tight_layout()
plt.show()
```

The strongest links to cancellation are lead time (+0.19), parking requested (−0.19), ADR (+0.13), special requests (−0.12) and guest-nights (+0.11).

All of them are weak (under 0.2). Cancellation has many causes, and no single variable predicts it.

Stay nights and guest-nights are 0.85 correlated because I built one from the other, so a model shouldn't use both.

The heatmap shows correlation, not causation, and only linear relationships.
